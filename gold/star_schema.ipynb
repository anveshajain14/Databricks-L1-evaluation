```sql
-- transaction fact table
create table identifier(:catalog || '.gold.fact_transaction') as
select t.transaction_id, t.account_id, t.transaction_date,t.transaction_timestamp, t.transaction_type, t.transaction_channel, t.amount, t.currency, t.merchant_name, t.merchant_category, t.transaction_status, t.reference_number, t.created_at, b.branch_id,a.interest_rate, a.customer_id
from identifier(:catalog || '.silver.silver_transactions') t 
join identifier(:catalog || '.silver.silver_accounts') a on t.account_id = a.account_id
join identifier(:catalog || '.silver.silver_branches') b on a.branch_id = b.branch_id
```

```sql
-- customer dimension table
create table identifier(:catalog || '.gold.dim_customer') as
select customer_id, first_name, last_name, date_of_birth, email, phone, address, city, state, postal_code, customer_segment, customer_status, registration_date, updated_at
from identifier(:catalog || '.silver.customers_scd')
```

```sql
-- accounts dimension tabel
create table identifier(:catalog || '.gold.dim_accounts') as
select account_id, account_type, account_status, opening_date, closing_date, currency, account_tier, updated_at
from identifier(:catalog || '.silver.silver_accounts')
```

```sql
-- branch dimension table
create table identifier(:catalog || '.gold.dim_branch') as
select branch_id, branch_code, branch_name, city, state, region, branch_type, manager_name, branch_status, opening_date
from identifier(:catalog || '.silver.silver_branches')
```
