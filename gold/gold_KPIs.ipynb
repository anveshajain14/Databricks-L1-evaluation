KPI 1 - Total Transaction Value 

```sql
create temporary view vw_total_transaction_value as
select sum(amount) as total_transcation_value 
from identifier(:catalog || '.gold.fact_transaction');

select * from vw_total_transaction_value
```

KPI 2 - Total Transaction Count

```sql
create temporary view vw_successful_transaction_count as
select count(transaction_id) as successful_transaction_count
from identifier(:catalog || '.gold.fact_transaction')
where transaction_status = "completed";

select * from vw_successful_transaction_count
```

KPI 3 - deposite vs withdrawals

```sql
select sum(amount) as total_deposits 
from identifier(:catalog || '.gold.fact_transaction')
where transaction_type = "deposit"
```

```sql
select sum(amount) as total_withdrawals 
from identifier(:catalog || '.gold.fact_transaction')
where transaction_type = "withdrawal"
```

KPI 4 - Monthly Transaction Value

```sql
select year(transaction_date) || '-' || month(transaction_date) as month, sum(amount) as total_monthly_transactions
from identifier(:catalog || '.gold.fact_transaction')
group by year(transaction_date), month(transaction_date) 
order by year(transaction_date), month(transaction_date)
```
