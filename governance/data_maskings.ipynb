```sql
create or replace function identifier(:catalog || ".silver.mask_email") (x string)
returns string
return concat(left(x,1), "***", substr(x,locate('@',x)))
```

```sql
create or replace function identifier(:catalog || ".silver.mask_phone") (x string)
returns string
return concat("******", right(x,4))
```

```sql
select customer_id, first_name, last_name, date_of_birth, 
case
when :user = 'DE' then email
else dev.silver.mask_email(email)
end
as email,
case
when :user = 'DE'
then phone
else dev.silver.mask_phone(phone)
end 
as phone,
address, city, state, postal_code, customer_segment, customer_status, registration_date, updated_at
from dev.gold.dim_customer
```
