```python
dbutils.widgets.text("catalog","dev")
catalog = dbutils.widgets.get("catalog")
```

```python
from pyspark.sql import functions as F
```

```python
df_branch = spark.read.table(f"{catalog}.bronze.bronze_branches")
```

```python
df_cleaned_branch = df_branch.withColumn("opening_date", F.to_date(F.col("opening_date"), "yyyy-MM-dd")) \
    .withColumn("city", F.lower(F.trim(F.col("city")))) \
        .withColumn("state", F.lower(F.trim(F.col("state")))) \
            .withColumn("branch_type", F.lower(F.trim(F.col("branch_type")))) \
                .withColumn("manager_name", F.trim(F.col("manager_name"))) \
                    .withColumn("branch_status", F.lower(F.trim(F.col("branch_status")))) \
                        .dropDuplicates() \
                            .dropna()
```

```python
df_cleaned_branch.write.mode("overwrite").saveAsTable(f"{catalog}.silver.silver_branches")
```
